## Step 3: First Look at the Data The dataset has 84 rows and 7 columns. Applicant Name, Email, University and Status each have 2 missing values. There are 3 rows that are exactly the same. The Application Date column is stored as text, not as a real date.

```python
df = df.dropna(subset=["Applicant Name"])
df = df.dropna(subset=["Email"])
df["University"] = df["University"].fillna("Not Specified")
df["Status"] = df["Status"].fillna("Under Review")
```

```python
df.isnull().sum()
```

```python
df.shape
```

## Step 4: Handling Missing Values

I removed the rows with a missing Applicant Name, because an application without a name cannot be used. I also removed the rows with a missing Email, because we need the email to contact the applicant.

I filled the missing University values with "Not Specified", because this information is not very important.

I filled the missing Status values with "Under Review", because it probably means the application has not been checked yet.

```python
df = df.drop_duplicates()
df.shape
```

```python
df["Email"] = df["Email"].str.lower()
df = df.drop_duplicates(subset=["Email"])
df.shape
```

## Step 5: Removing Duplicates

I found 3 rows that were exactly the same and removed them. Then I made all emails lowercase and removed 1 more row that had the same email as another row, but with the name written in capital letters.

I checked duplicates by email because a person can write their name in different ways, but their email stays the same. So email is a more reliable way to find the same person.

```python
df["Applicant Name"] = df["Applicant Name"].str.strip().str.title()
df["Status"] = df["Status"].str.strip().str.title()
```

```python
df["Domain Applied"] = df["Domain Applied"].str.strip().str.lower()

domain_fix = {
    "web dev": "Web Development",
    "web development": "Web Development",
    "data science": "Data Science",
    "ai/ml": "AI/ML",
    "app dev": "App Development",
    "app development": "App Development",
    "graphic design": "Graphic Design"
}

df["Domain Applied"] = df["Domain Applied"].map(domain_fix)
df["Domain Applied"].unique()
```

```python
df["University"] = df["University"].str.strip().str.lower()

uni_fix = {
    "nust": "NUST",
    "fast": "FAST",
    "uet": "UET",
    "pu": "Punjab University",
    "punjab university": "Punjab University",
    "not specified": "Not Specified"
}

df["University"] = df["University"].map(uni_fix)
df["University"].unique()
```

```python
df["Status"].unique()
```

## Step 6: Standardizing Text

The same value was written in different ways, for example "web dev" and "WEB DEV". The computer sees these as different values, so counts would be wrong.

I removed extra spaces and made the casing the same. Then I mapped all the different spellings of Domain Applied to 5 clean names. I did the same for University and Status.

```python
df["Application Date"] = pd.to_datetime(df["Application Date"], format="mixed", dayfirst=True, errors="coerce")
df["Application Date"].isnull().sum()
```

```python
df["Phone"] = df["Phone"].str.replace("-", "", regex=False)
df["Phone"] = df["Phone"].str.replace(" ", "", regex=False)
df["Phone"].head()
```

```python
df.info()
```

## Step 7: Fixing Data Types

The Application Date column was stored as text, so it could not be sorted or used for calculations. I converted it to a real date type. I used errors="coerce" so that any unreadable date would become empty instead of crashing the notebook. After the conversion, there were 0 empty dates.

The Phone column had dashes and spaces in different places. I removed them to make the format the same. I kept Phone as text so the leading zero does not disappear.

## Data Quality Summary

**Starting data:** 84 rows and 7 columns.

**Problems I found:**
- Some cells were empty: 2 names, 2 emails, 2 universities and 2 statuses.
- Some rows were repeated: 3 were exactly the same, and 1 more had the same email as another row.
- The same domain was written in many ways. There were 12 different spellings for only 5 real domains.
- University and Status were also written in different ways, like "nust" and "NUST".
- Application Date was saved as text, not as a real date.
- Phone numbers had dashes and spaces.

**What I did to fix them:**
- I removed the 2 rows with no name, because we cannot use an application without a name.
- I removed the 2 rows with no email, because we need the email to contact the person.
- I wrote "Not Specified" in the empty University cells, because it is not very important.
- I wrote "Under Review" in the empty Status cells, because the application was probably not checked yet.
- I removed 4 repeated rows in total: 3 exact copies and 1 with the same email.
- I made the spellings the same. Now there are only 5 domains and 3 statuses.
- I changed Application Date into a real date.
- I removed the dashes and spaces from the phone numbers.

**Final data:** 76 rows and 7 columns. There are no empty cells, no repeated rows, and all the text looks the same .
